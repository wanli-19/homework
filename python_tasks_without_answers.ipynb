
### 1. 简单的输入输出
**题目**: 使用`input()`函数，要求用户输入一个整数，输出其平方。

### 2. 判断闰年
**题目**: 输入一个年份，判断该年份是否为闰年。闰年的判定条件为：
- 如果年份能被4整除且不能被100整除，或者能被400整除，则为闰年。

### 3. 条件判断 - if语句
**题目**: 编写程序，输入一个整数，判断它是否为正数。

### 4. while循环
**题目**: 编写一个程序，要求用户输入一个正整数，计算1到该数之间所有整数的和。

### 5. for循环
**题目**: 打印1到10之间所有的偶数。

### 6. 分支语句
**题目**: 输入一个成绩，判断其所属的等级（A: >=90, B: >=80, C: >=70, D: >=60, E: <60）。

### 7. 嵌套if语句
**题目**: 输入一个整数，判断其是否为正奇数、正偶数、负奇数或负偶数。

### 8. while循环与break语句
**题目**: 编写一个程序，要求用户输入数字并计算和，直到输入负数时停止。

### 9. for循环与continue语句
**题目**: 打印1到20之间所有的奇数，跳过5的倍数。

### 10. 数字反转
**题目**: 输入一个三位数的整数，输出该数字的反转数。

### 11. 列表操作
**题目**: 输入5个整数，将其存入列表并计算其总和与平均值。

### 12. 字典操作
**题目**: 创建一个字典存储学生的姓名和成绩，允许用户查询某个学生的成绩。

### 13. 最大子数组和
**题目**: 给定一个整数数组，找出具有最大和的连续子数组。

### 14. 两数之和
**题目**: 给定一个整数数组和一个目标值，找到数组中和为目标值的两个数。

### 15. 计算阶乘
**题目**: 输入一个正整数，计算该数字的阶乘。
